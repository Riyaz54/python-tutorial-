STRING SLICING:- A string in python can be sliced for getting a part of the string 


```python
name = "MOHAMMAD RIYAZ CHOUHAN"
print(len(name))
```

```python
name = "MOHAMMAD RIYAZ CHOUHAN"

print(name[0:3])
```

```python
name = "MOHAMMAD RIYAZ CHOUHAN"
print(name[-4: -1])
print(name[1:4])
```

Slicing with skip value:- We can provide a skip value as a part of our slice like this  

```python
word = "Python"
word
print(type(word))
print(word[0:1:3])
```

Most used string functions in python:- 

| Method         | What it does                         | Example                                        |
| -------------- | ------------------------------------ | ---------------------------------------------- |
| `lower()`      | Converts to lowercase                | `"HELLO".lower()` → `"hello"`                  |
| `upper()`      | Converts to uppercase                | `"hello".upper()` → `"HELLO"`                  |
| `strip()`      | Removes whitespace from both ends    | `" hi ".strip()` → `"hi"`                      |
| `lstrip()`     | Removes whitespace from left         | `" hi".lstrip()` → `"hi"`                      |
| `rstrip()`     | Removes whitespace from right        | `"hi ".rstrip()` → `"hi"`                      |
| `replace()`    | Replaces text                        | `"hello".replace("h", "y")` → `"yello"`        |
| `split()`      | Splits into a list                   | `"a,b,c".split(",")` → `["a","b","c"]`         |
| `join()`       | Joins strings together               | `"-".join(["a","b","c"])` → `"a-b-c"`          |
| `find()`       | Finds position of substring          | `"hello".find("l")` → `2`                      |
| `count()`      | Counts occurrences                   | `"banana".count("a")` → `3`                    |
| `startswith()` | Checks beginning                     | `"hello".startswith("he")` → `True`            |
| `endswith()`   | Checks ending                        | `"file.pdf".endswith(".pdf")` → `True`         |
| `isdigit()`    | Checks if all characters are digits  | `"123".isdigit()` → `True`                     |
| `isalpha()`    | Checks if all characters are letters | `"hello".isalpha()` → `True`                   |
| `isalnum()`    | Checks if letters/numbers only       | `"abc123".isalnum()` → `True`                  |
| `isspace()`    | Checks if whitespace only            | `"   ".isspace()` → `True`                     |
| `capitalize()` | Capitalizes first character          | `"hello world".capitalize()` → `"Hello world"` |
| `title()`      | Capitalizes each word                | `"hello world".title()` → `"Hello World"`      |
