Operation in Tuple:- commonly taught operation 7 hote hai..!!

Python में Tuple के commonly taught operations कुल 7 माने जा सकते हैं:

Concatenation (+) → दो tuples जोड़ना

Repetition (*) → tuple को repeat करना

Membership (in) → element मौजूद है या नहीं

Membership (not in) → element मौजूद नहीं है या नहीं

Indexing ([]) → किसी position का element निकालना

Slicing ([:]) → tuple का एक हिस्सा निकालना

Comparison (==, !=, <, >, <=, >=) → दो tuples compare करना

साथ में याद रखो:

Tuple methods: 2 → count(), index()

Common operations: 7 categories ऊपर वाली

Tuple immutable है, इसलिए append(), remove(), insert() जैसे operations नहीं होते।

```python
#1.Concatenation:- Matlab two or two se jayda tuples ko add karne me use ho..! islite CONCATENATION ka use hota hai 
# Basicly 2 Tuples ko add karna..!! 
t1 = (1, 2, 3, 4, 5)
t2 = (6, 7, 8, 9, 11)

t3 = t1 + t2 
print(t3)
```

```python
#2. Repetition:- ka mtlb hai elements ko baar baar repat karna...!! iske liyee REPETTION ka use hiota hai 
# Tuples ko multiple times repat karna...!!

t1 = (1, 2, 3, 4, 5)
print(t1 * 2)
print(t1 * 3)
```

```python
name = ("Riyaz", "Sameer", "Sahil", "Jay")
print(name * 2)
print(name * 1)

#1000,2000>>> jitne chaiye utne times likh skte hai
```

```python
#3.Membership:- iska kaam check karna hota hai ki koi element tuple me mojud hai ya nhi 
# Iske liye 2 Operator hote hai..!!
#1. in:- Agr element hai to True
#2. not in:- Agr element nhi hai to Flase

name = ("Riyaz", "Sameer", "Sahil", "Jay")

print("riyaz" in name) #kyu ki riyaz small me likha hai jab ki tuple me Riyaz aise likha hai 
print("Riyaz" in name)

print("sana" in name)
print("sana" not in name)
```

```python
#5.Indexing:- ka mtlb tuple me kesi element ko uski position/index number me access karna 
# Index hamesha 0(zero) se shuru hoti hai 

name = ("Riyaz", "Sameer", "Sahil", "Jay")
print(name[0])
print(name[1])
print(name[2])
print(name[3])
```

ISME Negative Indexing bhi hota hai...!!!



```python
name = ("Riyaz", "Sameer", "Sahil", "Jay")
print(name[-3])
print(name[-1])
```

```python
#6.Slicing:- Ka mtlb tuple me kesi ek particular range ke element nikalna..!1

t = (10, 20, 30, 40, 50)
print(t[1 : 2])
print(t[2 : 3])
```

```python
N = ("Riyaz", "Sameer", "Sahil", "Jay")
print(N[1 : 2])
```

```python
#7. Comparsion:- Mtlb 2 tuple ko compare karna 
# ==,!=,<>,<=>=

t1 = (1, 2, 3)
t2 = (1, 2, 4)

print(t1 == t2)
print(t1 < t2)
```

```python
t3 = 22 + 45
t4 = 44 + 45 
print(t3)
print(t4)

print(t3 == t4)
print(t3 < t4)
print(t3 > t4)
```

```python
print(t3 == t4)  # Equal
print(t3 != t4)   #Not equal
print(t4 != t3)
print(t3 <= t4)   #Less than or equal
print( t3 >= t4)    #Greater than
```
