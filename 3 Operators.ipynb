Truth Table of OR 


```python
print("True or False is ", True or False)
print("True or True is ", True or True)
print("False or True is ", False or True)
print("False or False is ", False or False)
```

Truth Table of "AND"

```python
print("True and False is ", True or False)
print("True and True is ", True or True)
print("False and True is ", False or True)
print("False and False is ", False or False)
```

Truth Table of "NOT"

```python
print("True not False is ", True or False)
print("True not True is ", True or True)
print("False not True is ", False or True)
print("False not False is ", False or False)
```
